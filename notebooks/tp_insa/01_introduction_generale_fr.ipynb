### Commentaires

Les commentaires sont des morceaux de textes qui ne sont pas exécutés par le programme. Ils servent à apporter des informations complémentaires pour que le code soit plus compréhensible. En Python, on utilise le caractères `#` pour commencer un commentaire, qui continuera jusqu'au bout de la ligne.

Exemple :
```Python
# La Télédétection c'est vraiment super

teledetection = "Super"  # Je vous l'avais bien dit
```

Il faut utiliser les commentaires intelligemment. S'il n'y a pas de commentaires du tout dans le code, une personne tierce (ou le vous de dans trois mois) n'aura probablement pas le courage de se plonger dans votre code pour comprendre ce que vous avez bien voulu faire. A contrario, s'il y a besoin de trop de commentaires, c'est que le reste du code n'est pas assez lisible. Il faut également veiller à mettre à jour les commentaires à jour quand on met son code à jour : mieux vaut pas de commentaire qu'un commentaire qui contredit le code. 

Les commentaires devraient être des phrases complètes commençant par une majuscule.

Les blocs de commentaires sont généralement composés d'un ou plusieurs paragraphes avec des phrases complètes se terminant par des points. On peut séparer les paragraphes avec une ligne qui est simplement un `#`.

On essaie de limiter les commentaires en ligne (ceux qui sont à la suite d'un morceau de code).

Exemple
```Python
# La Télédétection c'est vraiment super

teledetection = "Super"  # Commentaire en ligne

# Maintenant on peut écrire un bloc de code sur le fait que le SERTIT c'est vraiment super.
# Là on fait une deuxième phrase mais bon c'est pour que ça fasse paragraphe quoi.
#
# En conclusion : le SERTIT c'est vraiment super. 
```

Lorsqu'on écrit un code qui a vocation à être potentiellement partagé à des gens à l'international, on écrit les commentaires en anglais.

## Listes

En Python, les listes sont une collection ordonnées d'éléments qui permettent de stocker et de manipuler efficacement des données.

Elles sont indiquées par des crochets `[ ]` et les éléments sont séparés par des virgules.

Exemple :

```Python
pixel_rgbnir = [125, 125, 125, 125]
```

Les listes étant des structures de données ordonnées, on peut accéder à un élément grâce à sa position dans la liste.

A noter que comme dans beaucoup de langages de programmation, les indices commencent à 0 et non à 1. Ici, notre quatrième élément est donc à l'indice 3. La conséquence de cette indentation est aussi que le dernier élément aura toujours l'indice `n-1` avec `n` la longueur de la liste : `list[n]` sera donc en-dehors de la liste.

Exemple :

```Python
r = pixel_rgbnir[0]
b = pixel_rgbnir[2]
```

### Manipulations de listes

Un des avantages des listes c'est qu'elles sont *mutables* (ou "muables" en français, mais personne ne dit ça), c'est-à-dire qu'on peut les modifier. On peut changer la valeur d'un élément, en ajouter, en enlever...

Pour ajouter un élément, on peut utiliser `append` comme ceci :

````python
pixel_rgbnir.append(0)
print(pixel_rgbnir)
````

Et l'enlever avec remove :

````python
pixel_rgbnir.remove(0)
print(pixel_rgbnir)
````

On peut aussi ajouter des listes entre elles avec l'opérateur `+`. Par exemple :

````python
pixel_rgb = [37, 150, 190]
pixel_nir = [0]

pixel_rgbnir = pixel_rgb + pixel_nir 
print(pixel_rgbnir)
````

Quelques manipulations utiles pour les listes :
- Le slicing : Plutôt que de récupérer simplement un élément avec `[]`, on peut récupérer toute une sous-partie d'une liste avec la syntaxe suivante : `liste[debut:fin:pas]`. Par défaut debut est le premier élément de la liste, fin le dernier. Il est facultatif de rajouter le `:pas`, et le pas vaut 1 par défaut. On peut ainsi récupérer toute la liste avec `liste[:]`, les éléments après `i` avec `liste[i:]`, les éléments avant `j` avec `liste[:j]`, et les éléments entre i et j avec `liste[i:j]`. `list[i:j:2]` récupère un élément sur 2 entre i et j.
- `len(liste)` donne la longueur de la liste `liste`
- `liste.count(element)` compte le nombre de "element" dans la liste
- `liste.sort()` renvoie la liste triée
- `liste.pop()` récupère le dernier élément de la liste et le supprime de la liste. Par exemple :

````python
pixel_nir = pixel_rgbnir.pop()
print(f'On récupère le pixel NIR : {pixel_nir} et le pixel RGB qui a perdu son NIR : {pixel_rgbnir}')
````

### Les autres types de données structurées

#### Tuples ou N-uplets

Les tuples sont un ensemble de valeurs Python qui sont présentées entre parenthèses. On peut y accéder avec des crochets comme les listes, mais contrairement aux listes, les tuples sont immuables.

#### Dictionnaires

Les dictionnaires en Python sont une implémentation d'objets clé-valeurs avec des clefs indexées. Les clefs sont le plus souvent des strings. Si vous avez déjà utilisé des fichiers `json` la syntaxe et la logique des dictionnaires ne devrait pas trop vous perturber : on écrit un dictionnaire entre accolades `{}`, en séparant la clef de la valeur par deux points, et en séparant chaque objet par une virgule.

Exemple :

````python
mon_point = {
    "type" : "Point",
    "coordonnees" : [-71, 41]
    }
````

On peut très simplement extraire des informations du dictionnaire, un peu comme avec une liste mais en renseignant la clef plutôt que l'indice : 

````python
mon_point["coordonnees"]
````

Petites informations en plus :
- `"clef" in mon_dictionnaire` permet de vérifier si "clef" est dans "mon_dictionnaire" en renvoyant un booléen
- `mon_dictionnaire.get("clef")` fait la même chose que l'accès par crochets, mais renvoie un élément vide `None` si la clef n'est pas dans le dictionnaire, au lieu de renvoyer une erreur.
- `mon_dictionnaire.keys()` renvoie les clefs de mon_dictionnaire
- `mon_dictionnaire.values()` renvoie les valeurs du dictionnaire
- `mon_dictionnaire.items()` renvoie les clefs et les valeurs sous forme de tuple

## Boucles

Les boucles permettent de répéter plusieurs fois un bloc d'instructions.

### Boucle while

Une boucle **while** exécute l'instruction **tant que** la condition est vérifiée.

Par exemple :

```Python
i = 0
while i < 5:
    i = i + 1
    print(i)
```

On verra en sortie :
`1 2 3 4 5 `

### Boucle for

La boucle **for** sert à itérer sur une séquence (une liste ou un tuple).

Par exemple :

````python
pixel_rgbnir = [120, 100, 110, 20]

for color in pixel_rgbnir :
    print(color)
````

Un usage courant des boucles **for** est de vouloir exécuter notre instruction un nombre connu *n* de fois. On utilisera à ce moment l'instruction `range(n)`, qui créé un tuple d'entiers de 0 à n-1 (avec la même logique que les listes, où le dernier élément a l'indice `n-1` quand `n` est la longueur de la liste).

Par exemple :

```Python
for i in range (5)
    print(i)
```

On verra en sortie :
`0 1 2 3 4`

Pour définir une fonction, on utilise le mot-clef def, suivi du nom de la fonction, de parenthèses contenant éventuellement les arguments puis de deux points. On peut ensuite faire appel à la fonction avec le nom de la fonction, puis de parenthèses contenant ou non des arguments.

````python
def bravo_la_teledec():
    print("La télédétection c'est vraiment super !")

bravo_le_sertit()
````

````python
def bravo_le_(nom_entreprise):
    print("Le", nom_entreprise, "c'est vraiment super !")

bravo_le_("sertit")
````

A noter que si on définit des variables de manière interne à la fonction, à moins de les renvoyer avec `return` elles ne sortiront jamais de la fonction. Quelques exemples pour comprendre : 

````python
def bravo_le_(nom_entreprise="sertit"):
    nom_entreprise_caché = "laboratoire icube"
    print("Le", nom_entreprise, "c'est vraiment super !")

bravo_le_()
print(nom_entreprise_caché)
````

On obtient une erreur (c'est normal ! `nom_entreprise_caché` n'existe que dans la fonction). Ca vaut même si le nom de variable est déjà utilisé autre part :

````python
nom_entreprise="sertit"

def bravo_le_():
    nom_entreprise = "laboratoire icube"

print("Le", nom_entreprise, "c'est vraiment super !")
````

Si en revanche on _veut_ modifier cette variable, il faut la renvoyer avec `return`

````python
nom_entreprise="sertit"

def bravo_le_(nom_entreprise):
    nom_entreprise = "laboratoire icube"
    return nom_entreprise

nom_entreprise = bravo_le_(nom_entreprise)
print("Le", nom_entreprise, "c'est vraiment super !")
````

Un des grands intérêts des fonctions est de nous éviter d'écrire plusieurs fois un morceau de code similaire ou presque. Elles permettent aussi de garder notre code court, concis et lisible, surtout en choisissant des noms de fonctions clairs. Pour avoir un bon nom de fonction, il faut le plus possible respecter la philosophie de *une fonction = une tâche*. Si votre fonction fait trop de choses pour qu'un nom de fonction décrive tout efficacement, c'est qu'il vous faut plusieurs fonctions.

Les fonctions permettent ainsi de fragmenter notre code, ce qui en plus d'être lisible est beaucoup plus pratique à maintenir et à modifier.

#### Axis
Beaucoup de fonctions prennent un argument "axis". 

Les axes en Numpy sont comme les axes dans un système de coordonnées. Pour une matrice à deux dimensions ce sera par exemple la "direction" des colonnes et la "direction" des lignes. En Numpy, l'axe "0" est la direction le long des lignes, et l'axe "1" est le long des colonnes.
![axes](../assets/numpy-axes.webp)
Quand le paramètre "axis" est appelé dans une fonction, le fonctionnement n'est pas forcément intuitif, mais l'axe donné sera celui sur lequel on va appliquer la fonction, et donc qui va être "écrasé". Le plus simple pour comprendre est de tester, par exemple avec la fonction somme :

````python
import numpy as np

# Création d'une fonction de 2x3
m = np.arange(0, 6).reshape([2,3])
print(m)
````

On a donc une matrice avec 2 lignes et 3 colonnes. Regardons la somme avec `axis=0`, donc dans la direction des lignes :

````python
s1 = np.sum(m, axis=0)
s1
````

On a donc donné l'axe "des lignes", mais on ne fait pas la somme des lignes, on applique la fonction _dans la direction des lignes_ et cet axe disparaît.

Si on essaie avec `axis=1` :

````python
s2 = np.sum(m, axis=1)
s2
````

Notons que lorsqu'on parle de la "shape"/forme d'une matrice, il s'agit de la longueur des axes et on suit la même logique de numérotations, donc pour une matrice à deux dimensions de shape (n, m), n sera le nombre de lignes (longueur de l'axe 0) et m m le nombre de colonnes (longueur de l'axe 1).

Maintenant que la question des axes est claire, passons aux fonctions Numpy en elles-mêmes.

#### Pour la création d'arrays :
- [`np.empty`](https://numpy.org/doc/stable/reference/generated/numpy.empty.html#numpy.empty) : créé une array vide selon la taille qu'on lui donne
- [`np.zeros`](https://numpy.org/doc/stable/reference/generated/numpy.zeros.html) : créé une array remplie de zeros selon la taille qu'on lui donne
- [`np.ones`](https://numpy.org/doc/stable/reference/generated/numpy.ones.html) : créé une array remplie de 1 selon la taille qu'on lui donne
- [`np.empty_like`](https://numpy.org/doc/stable/reference/generated/numpy.empty.html#numpy.empty_like) : créé une array vide avec la même taille qu'une autre matrice donnée en argument
Il existe aussi [`np.zeros_like`](https://numpy.org/doc/stable/reference/generated/numpy.zeros_like.html) et [`np.ones_like`](https://numpy.org/doc/stable/reference/generated/numpy.ones_like.html)

#### Pour la manipulation d'arrays :
- [`np.sort`](https://numpy.org/doc/stable/reference/generated/numpy.sort.html#numpy.sort) : trie une array selon un axe
- [`np.concatenate`](https://numpy.org/doc/stable/reference/generated/numpy.concatenate.html#numpy-concatenate) : ajoute plusieurs arrays entre elles suivant un axe.
 - [`np.reshape`](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html#numpy-reshape) : Donne une nouvelle forme à une array sans changer son contenu, par exemple `np.reshape([[0, 0, 0], [0, 0, 0]], (3, 2))` renvoie `[[0, 0], [0, 0] ,[0, 0]]`. La nouvelle `shape` doit bien sûr être compatible avec la `shape` de base de notre array. Si on veut qu'une dimension soit déduite des autres, on peut mettre "-1", par exemple `np.reshape([[0, 0, 0], [0, 0, 0]], (-1, 2))` qui déduira que pour avoir 2 colonnes pour une matrice de 6 éléments, il faut 3 lignes. 
- [`np.where`](https://numpy.org/doc/stable/reference/generated/numpy.where.html) : sélectionne des éléments de matrices selon une condition. 
On appelle `np.where` de la manière suivante : `np.where(condition, x, y)`, et le retour est une matrice de la même forme que "condition", qui prend ses éléments depuis _x_ ou depuis _y_ selon si la condition est vérifiée ou non. Techniquement, tous les paramètres doivent être des matrices de la même taille, mais la fonction s'arrange pour que tous les cas de figure fonctionnent. Notamment :
    - L'instruction `a > 3` où `a` est une matrice renvoie en fait une matrice de booléens, avec `True` là où la condition est vérifiée et `False` sinon.
    - Si on donne simplement `x=1` (par exemple), la fonction va travailler avec une matrice de `1` de la même taille que `condition`
Le fonctionnement "classique" attendu ressemble ainsi à :
```Python
np.where([[True, False], [True, True]],
         [[1, 2], [3, 4]], [[5, 6], [7, 8]])
```
Qui retourne :
```Python
array([[1, 6],
       [3, 4]])
```
Mais on peut également faire par exemple : 
```Python
np.where([[1, 2],
          [3, 4]],
         [[5, 6],
          [7, 8]] > 3, 4, 0)
```
Qui retourne :
```Python
array([[0, 0], 
       [0, 4]],
      [[4, 4],
       [4, 4]])
```
**Bonus** : Si on ne fournit pas x et y, `np.where` renvoie les indices où la condition est vérifiée.
